```python
# Import Path to work with folders and file paths
from pathlib import Path

# Import pandas for data manipulation and analysis
import pandas as pd

# Locate the raw data folder relative to this notebook
data_dir = Path(r"C:\Users\Erick.Odhiambo\Documents\Projects\Football-Jersey-Sales-RFM-Customer-Segmentation\01_data\raw")

# Find all CSV whose names start with 'sales'
sales_csv_files = sorted(data_dir.glob("sales*.csv"))

# Display the number of files and their names
print(f"Number of CSVs found: {len(sales_csv_files)}")

for file in sales_csv_files:
    print(file.name)
```

```python
# Create an empty list to hold the monthly DataFrames
sales_frames = []

# Read each CSV file into pandas
for file in sales_csv_files:

    # Load the current CSV file
    df = pd.read_csv(file)

    # Record the source filename in a new column for traceability
    df["source_file"] = file.name

    # Add/append the DataFrame to our list
    sales_frames.append(df)

    # Display basic information about each file
    print(f"{file.name}: {df.shape[0]} rows, {df.shape[1]} columns")

# Combine all monthly DataFrames into one DataFrame
sales_df = pd.concat(sales_frames, ignore_index=True)

# Display the shape of the combined DataFrame
print("\nCombined dataset:")
print(f"Rows: {sales_df.shape[0]:,}")
print(f"Columns: {sales_df.shape[1]}")
```
